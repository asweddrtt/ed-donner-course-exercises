## Steps

1. **Gemini client**: load the key from `.env`, create the client with the Gemini base URL
2. **System prompt**: a patient Python & LLM-engineering tutor who explains code line by line
3. **Question**: store it in a variable, e.g.
   ```
   Please explain what this code does and why:
   yield from {book.get("author") for book in books if book.get("author")}
   ```
4. **Messages list**: system prompt + question
5. **Streaming answer** from `gemini-3.1-flash-lite`: print the text as it arrives (`print(..., end="")` in the loop)
6. **Normal answer** from `gemini-4-argon`: no streaming, then compare it with step 5
7. **Bonus (JSON)**: ask for JSON with keys `"summary"`, `"explanation"`, `"difficulty"` (beginner / intermediate / advanced) → convert with `json.loads` → print only the difficulty

## Reminders (only if stuck)

- Streaming: `stream=True`, then the text is in `chunk.choices[0].delta.content`
- JSON: `response_format={"type": "json_object"}` + an example of the format in the prompt

````python
import os
import requests
import json
from dotenv import load_dotenv

from openai import OpenAI
````

````python
load_dotenv(override=True)
gemini_api_key = os.getenv("GEMINI_API_KEY")
gemini_base_url = os.getenv("GEMINI_BASE_URL")

gemini = OpenAI(
    base_url=gemini_base_url,
    api_key=gemini_api_key
)

model = "gemini-3.1-flash-lite"
model0 = "gemini-3.5-flash-lite"

messages = [
    {
        "role" : "system",
        "content" : "You are a patient Python & LLM-engineering tutor who explains code line by line."
    },
    {
        "role" : "user",
        "content" : "Please explain what this code does and why:\n\n yield from {book.get('author') for book in books if book.get('author')}"
    }
]
````

````python
respone = gemini.chat.completions.create(
    model=model,
    messages=messages,
    stream=True
)
for chunk in respone:
    print(chunk.choices[0].delta.content, end="", flush=True)
````

````python
respone = gemini.chat.completions.create(
    model=model0,
    messages=messages
)
print(respone.choices[0].message.content)
````

````python
messages = [
    {
        "role" : "system",
        "content" : "You are a patient Python & LLM-engineering tutor who explains code line by line. reply in JSON format with keys 'summary', 'explanation', 'difficulty' (beginner / intermediate / advanced) like this: {""summary"" : this is about bla bla bla,""difficulty : ""beginner""}"
    },
    {
        "role" : "user",
        "content" : "Please explain what this code does and why:\n\n yield from {book.get('author') for book in books if book.get('author')}"
    }
]
````

````python
response = gemini.chat.completions.create(
    model=model0,
    messages=messages,
    response_format={
        "type": "json_object"
    }
)
data = json.loads(response.choices[0].message.content)
print(data["difficulty"])

````
